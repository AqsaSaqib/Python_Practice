```python
# What does .apply() do and how is it different from vectorized operations?
"""
.apply() function is used to perform complex mathemeatical operations on the data-frame and it works like a loop means that
all the operations must be performed one by one on each element instead of applying at once. and it takes more time to execute
On the other hand vectorized operations are used to perform simple arithmetic opertions on the data-frame. all the operations
are applied at once and it takes less time to execute
"""
```

```python
# Difference between .pivot() and .pivot_table()?
"""
Both .pivot() and .pivot_table() are used to reshape the data-frame from long format to wide format
The main difference between them is that the .pivot() doesn't handles any error and doesn't perform any aggregation function
but the .pivot_table() calculates the mean value nd fill all the NaN values with them and aggregate function also applies on it.
it handles error efficiently and performs aggregation function and at last provides the summarize data.
"""
```

```python
# What does .merge() do and what are the 4 types of joins?
"""
.merge() is used to merge or join two data-frame with one another and the four types of joins are:
Left join, Right join, Inner join and Outer join
"""
```

# **Implementation Task**

```python
import pandas as pd
```

# **Load DataSet**

```python
df = pd.read_csv("/content/Superstore.csv", encoding='latin1')
df.head(5)
```

# **DataSet Information**

```python
print("Shape:\n", df.shape)
print()
print("Columns:\n", df.columns)
print()
print(df.info())
print()
print("Statistics:\n", df.describe())
```

# **Check Missing values**

```python
print(df.isnull().sum())
```

```python
print("There is no Missing Value in the dataset")
```

# **Fix any wrong datatype**

```python
print("Before conversions:\n", df.dtypes)
```

```python
# Convert date columns to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Convert Postal Code to string
df["Postal Code"] = df["Postal Code"].astype("string")

print("After conversion:\n", df.dtypes)
```

```python
# Check for missing dates
date_missing = df[["Order Date", "Ship Date"]].isnull().sum()
print("Missing values in date columns:\n", date_missing)
```

# **Drop Duplicates**

```python
# Count duplicates
duplicate_count = df.duplicated().sum()
print("Duplicate rows before cleaning:", duplicate_count)
```

```python
# Remove duplicates
df = df.drop_duplicates()
print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New dataset shape:", df.shape)
```

# **Derived Columns**

```python
# Calculate shipping days
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"])
# Display result
display(df[["Order Date", "Ship Date", "Shipping Days"]].head())
```

```python
# Calculate profit margin
df["Profit Margin"] = df["Profit"] / df["Sales"]
# Display result
display(df[["Sales", "Profit", "Profit Margin"]].head())
```

# **Verfiying Final Data-Frame**

```python
print("Total Missing Values:")
print(df.isnull().sum())

print("\nTotal Duplicate Rows:")
print(df.duplicated().sum())

print("\nOrder Date Data Type:")
print(df["Order Date"].dtype)

print("\nShip Date Data Type:")
print(df["Ship Date"].dtype)

print("\nDerived Columns:")
print("Shipping Days" in df.columns)
print("Profit Margin" in df.columns)
```

# **Summary Table after groupby and aggregate**

```python
summary = (df.groupby("Category").agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Average_Profit_Margin=("Profit Margin", "mean")))
# Sort by total sales
summary = summary.sort_values("Total_Sales", ascending=False)
display(summary)
```

# **New Small DataFrame**

```python
category = pd.DataFrame({"Category": ["Furniture","Office Supplies","Technology"],"Business_Area": ["Workspace & Home","Office Essentials","Technology Products"]})
display(category)
```

# **Merge the DataFrame**

```python
final_summary = summary.merge(category, on="Category",how="left")
display(final_summary)
```

# **Final DataSet**

```python
print("Final cleaned dataset shape:", df.shape)
print("\nFinal columns:")
print(df.columns)
print("\nFinal Summary Table:")
display(final_summary)
```
