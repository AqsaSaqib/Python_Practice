# **Bussiness Questions**

```python
# Q no.1 Which products are generating the highest profits?
q1 = df[df["Profit"] > 100][["Product Name", "Sales", "Quantity", "Profit"]]
q1
```

```python
# Q no.2 Which products generate less profit?
q2 = df[df["Profit"] < 0][["Product Name", "Category", "Sub-Category", "Sales", "Discount", "Profit"]]
q2
```

```python
# Q no.3 Which Order ID have a high discount and negative profit?
q3 = df[(df["Discount"] >= 0.50) &(df["Profit"] < 0)][["Order ID", "Product Name", "Sales", "Discount", "Profit"]]
q3
```

```python
# Q no.4 Which products have high sales but negative profit?
q4 = df[(df["Sales"] > 1000) & (df["Profit"] < 0)][["Product Name", "Sales", "Discount", "Profit"]]
q4
```

```python
# Q no.5 Which regions have sales transactions with negative profit?
q5 = df[df["Profit"] < 0][["Region", "Product Name", "Sales", "Discount", "Profit"]]
q5
```

```python
# Q no.6 Which State generate the high profit?
q6 = df.groupby("State")["Profit"].sum().sort_values(ascending=False)
q6
```

```python
# Q no. 7 Which product and subcategory generate the highest sales?
q7 = df.groupby(["Product Name", "Sub-Category"])["Sales"].sum().sort_values(ascending=False)
q7
```

```python
# Q no. 8 Which category and sub-category transactions are creating loss?
q8 = df[df["Profit"] < 0][["Category", "Sub-Category", "Product Name","Sales", "Discount", "Profit"]]
q8
```
