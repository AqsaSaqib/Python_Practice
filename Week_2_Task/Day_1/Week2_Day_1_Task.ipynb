# **Players_Info Cleaning**

```python
players_before_cleaning = len(players_info)
print("Rows before cleaning:", players_before_cleaning)
```

**Clean Columns Names**

```python
players_info.columns = players_info.columns.str.strip().str.lower()
print("Cleaned column names:\n", players_info.columns)
```

**Player ID Cleaning**

```python
players_info["id"] = pd.to_numeric(players_info["id"],errors="coerce")
```

```python
# Check Missing IDs
print("Missing IDs after conversion:\n",players_info["id"].isnull().sum())
```

```python
# Remove rows which has missing ids
players_info = players_info.dropna(subset=["id"])
```

**Remove Duplicates**

```python
duplicates_removed = players_info.duplicated().sum()
```

```python
players_info = players_info.drop_duplicates()
print("Duplicate rows removed:\n", duplicates_removed)
```

**Repeated Player IDs**

```python
print("Duplicate Player IDs:",players_info["id"].duplicated().sum())
```

**Date columns**

```python
date_columns = ["born_date","debut_date","last_date"]
```

```python
for col in date_columns:
    players_info[col] = pd.to_datetime(players_info[col],errors="coerce")
```

```python
print("\nDate column data types:")
print(players_info[date_columns].dtypes)
```

```python
print("Rows before cleaning:", players_before_cleaning)
print("Rows after cleaning:", len(players_info))
print("Rows removed:", players_before_cleaning - len(players_info))
```

```python
# Missing values
print("Missing values:\n",players_info.isnull().sum())
```

```python
# Duplicate rows
print("\nDuplicate rows:", players_info.duplicated().sum())
```

**Handle Missing Values**

```python
# Missing profile pictures
players_info["profile_pic"] = players_info["profile_pic"].fillna("No Image")

# Missing common names
players_info["player_common_names"] = players_info["player_common_names"].fillna("Unknown")

# Missing teams
players_info["player_teams"] = players_info["player_teams"].fillna("Unknown")
```

```python
print("Missing values after cleaning:\n", players_info.isnull().sum())
print("\nDuplicate rows after cleaning:\n", players_info.duplicated().sum())
```

# **Cleaned Players_Info DataSet**

```python
players_info.to_csv("/content/cleaned_players_info.csv",index=False)
print("\nCleaned dataset saved successfully!")
```

# **Seasonal_Stats Cleaning**

```python
seasonal_before_Cleaning = len(seasonal_stats)
print("Rows before cleaning:", seasonal_before_Cleaning)
```

**Clean columns Names**

```python
seasonal_stats.columns = seasonal_stats.columns.str.strip().str.lower()
print("Cleaned column names:\n", seasonal_stats.columns)
```

**Player ID Cleaning**

```python
seasonal_stats["player_id"] = pd.to_numeric(seasonal_stats["player_id"],errors="coerce")
```

```python
print("Missing player IDs:",seasonal_stats["player_id"].isnull().sum())
```

```python
seasonal_stats = seasonal_stats.dropna(subset=["player_id"])
```

**Remove Duplicates**

```python
seasonal_duplicates = seasonal_stats.duplicated().sum()
seasonal_stats = seasonal_stats.drop_duplicates()
print("Duplicate rows removed:\n",seasonal_duplicates)
```

**Make Year as Numeric Column**

```python
seasonal_stats["year"] = pd.to_numeric(seasonal_stats["year"],errors="coerce")
print("Missing year values:",seasonal_stats["year"].isnull().sum())
```

**Numeric Columns Sepration**

```python
numeric_columns = seasonal_stats.select_dtypes(include=["number"]).columns
print("Numeric columns:",numeric_columns.tolist())
```

**Missing Values in Numeric Columns**

```python
print("\nMissing values in numeric columns:")
print(seasonal_stats[numeric_columns].isnull().sum())
```

```python
# Columns which are not included
exclude_columns = ["player_id","year"]
```

```python
for col in numeric_columns:
    if col not in exclude_columns:
        if seasonal_stats[col].isnull().sum() > 0:
            median_value = seasonal_stats[col].median()
            seasonal_stats[col] = seasonal_stats[col].fillna(median_value)
            print(f"{col}: missing values filled with median = {median_value}")
```

**After Handling Missing Values**

```python
missing_after = seasonal_stats.isnull().sum()
print(missing_after)
```

```python
total_missing = seasonal_stats.isnull().sum().sum()
print("Total remaining missing values:\n",total_missing)
```

**Final Evaluation of Missing and Duplicates**

```python
print("Rows before cleaning:", seasonal_before_Cleaning)
print("Rows after cleaning:", len(seasonal_stats))
print("Rows removed:", seasonal_before_Cleaning - len(seasonal_stats))
```

```python
print("Missing values after cleaning:\n",seasonal_stats.isnull().sum())
print("\nDuplicate rows after cleaning:\n",seasonal_stats.duplicated().sum())
```
