# **Conceptual Questions**

```python
# When should you use a bar chart vs. a histogram?
"""
A bar chart is used when we have to compare categorical values such as Clothes, mobiles, Electronics categories or much more
and we use a histogram to show the distiribution of numerical values such as distribution of students, marks, ages etc.
"""
```

```python
# What does a boxplot show that a histogram doesn’t?
"""
A box plot shows different things such as: Median, Q1, Q3, Outliers. and it make easy to comapre median, outliers and normal in range values.
on the other hand histogram just shows the distribution after changing the range of values.
"""
```

```python
# How do you add labels, title, and legend to a Matplotlib chart?
"""
we can add title, labels and legend in matplotlib by using:
plt.title()
plt.xlabel()
plt.ylabel()
plt.legend()
"""
```

```python
# What’s a heatmap typically used for in EDA?
"""
A heatmap is used to show the correlation between different numerical values.
Correlation values usually range from -1 to +1:
+1 shows Strong positive relationship
0 shows No relationship
-1 shows strong negative relationship
"""
```

```python
# Why is a pie chart usually a bad choice for data with more than 4–5 categories?
"""
When we have more than 4 or 5 categories, it is difficult to read the Pie chart because there are too many small portions
and we can't clearly read the pie chart which slice is bigger so it is better to use bar chart in this condition.
"""
```

# **Implementation Task**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/merged_players.csv")
df
```

# **Graphs Visulaization**

***Line Chart***

```python
yearly_score = df.groupby("year")["avg_score"].mean().reset_index()
plt.figure(figsize=(10, 5))
plt.plot(yearly_score["year"],yearly_score["avg_score"],marker="o",linewidth=2)
plt.title("Average Player Score by Year", fontsize=15)
plt.xlabel("Year")
plt.ylabel("Average Score")
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

This chart shows that the avg player score has changed across different years

***Bar Chart***


```python
team_score = (df.groupby("team")["total_score"].sum().sort_values(ascending=False).head(10).reset_index())
plt.figure(figsize=(12, 6))
sns.barplot(data=team_score,x="team",y="total_score",palette="rocket")
plt.title("Top 10 Teams by Total Score")
plt.xlabel("Team", fontsize=12)
plt.ylabel("Total Score", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

This chart compares the total scoring performance of the top 10 teams in the dataset.

***Histogram***

```python
plt.figure(figsize=(10, 6))
sns.histplot(df["avg_fantasy_points"],bins=30,kde=True,edgecolor="black",color="skyblue")
plt.title("Distribution of Average Fantasy Points")
plt.xlabel("Average Fantasy Points", fontsize=12)
plt.ylabel("Number of Players", fontsize=12)
plt.show()
```

This histogram shows how average fantasy points are distributed among the players

***BoxPlot***

```python
top_teams = df["team"].value_counts().head(8).index
boxplot_data = df[df["team"].isin(top_teams)]
plt.figure(figsize=(12, 6))
sns.boxplot(data=boxplot_data,x="team",y="avg_score",palette="Set2")
plt.title("Average Score Distribution by Team")
plt.xlabel("Team", fontsize=12)
plt.ylabel("Average Score", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

This boxplot compares the median, spread, and possible outliers of player scores across team

***Scatter Plot***

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,x="avg_disposals",y="avg_fantasy_points",alpha=0.6,color="teal")
plt.title("Average Disposals vs Average Fantasy Points")
plt.xlabel("Average Disposals", fontsize=12)
plt.ylabel("Average Fantasy Points", fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

This chart shows whether players with more average disposals tend to achieve higher fantasy points.

***Heatmap***

```python
performance_columns = [
    "avg_kicks",
    "avg_marks",
    "avg_handballs",
    "avg_disposals",
    "avg_goals",
    "avg_tackles",
    "avg_clearances",
    "avg_inside_50s",
    "avg_contested_possessions",
    "avg_uncontested_possessions",
    "avg_score",
    "avg_fantasy_points",
    "avg_percentage_played"
]
correlation = df[performance_columns].corr()
plt.figure(figsize=(14, 10))
sns.heatmap(correlation,annot=True,fmt=".2f",cmap="coolwarm",linewidths=0.5,square=True)
plt.title("Correlation Between Player Performance Metrics")
plt.show()
```

This heatmap shows which player performance metrics have strong positive or negative relationships with each other.
