```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/merged_players.csv")
df.head(10)
```

# **Q no.1 : Which teams have played the most games?**

```python
team_games = df.groupby("team")["games_played"].sum()
team_games = team_games.sort_values(ascending=False)
print(team_games.head(10))
```

**Graph Visualization**

```python
team_games.head(10).plot(kind="bar")
plt.title("Top 10 Teams by Games Played")
plt.xlabel("Team")
plt.ylabel("Total Games")
plt.show()
```

**Bussiness Insight**

```python
print("Business Insight:")
print("Teams with more games can be studied for experience and performance.")
```

**Observations**

```python
# Observations
print("Observations:")
print("1. The team with the highest total games has the most game activity.")
print("2. The chart compares the top 10 teams.")
print("3. Teams with more games have greater historical participation.")
```

# **Q no.3 : Which player teams have the highest number of players?**

```python
player_teams = df["player_teams"].value_counts()
print(player_teams.head(10))
```

**Graph Visualization**

```python
player_teams.head(10).plot(kind="bar")
plt.title("Top 10 Player Teams")
plt.xlabel("Player Teams")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.show()
```

**Bussiness Insight**

```python
print("Business Insight:")
print("Player-team counts help understand team representation and composition.")
```

**Observations**

```python
print("Observations:")
print("1. The chart shows the most common player-team groups.")
print("2. Some player-team groups have more records than others.")
print("3. The top group has the highest number of records.")
```

# **Q no.4 : How does player weight vary across teams?**

```python
team_weight = df.groupby("team")["weight"].mean()
team_weight = team_weight.sort_values(ascending=False)
print(team_weight)
```

**Graph Visualization**

```python
team_weight.iloc[:5].plot(kind="bar")
plt.title("Average Player Weight for First 5 Teams")
plt.xlabel("Team")
plt.ylabel("Average Weight (kg)")
plt.xticks(rotation=45)
plt.show()
```

**Bussiness Insight**

```python
print("Business Insight:")
print("Weight analysis can support fitness planning and player recruitment.")
```

**Observations**

```python
print("Observations:")
print("1. Average weight is different across teams.")
print("2. Some teams have higher average player weight.")
print("3. The chart compares physical characteristics of teams.")
```

# **Q no.6 : Who are the Top 15 goal scorers?**

```python
goal_scorers = df.groupby("player_name")["goals"].sum()
goal_scorers = goal_scorers.sort_values(ascending=False)
top_15 = goal_scorers.head(15)
print(top_15)
```

**Graph Visualization**

```python
top_15.sort_values().plot(kind="barh")
plt.title("Top 15 Goal Scorers")
plt.xlabel("Total Goals")
plt.ylabel("Player")
plt.show()
```

**Bussiness Insight**

```python
print("Business Insight:")
print("Top goal scorers help teams identify valuable attacking talent.")
```

**Observations**

```python
print("Observations:")
print("1. The chart shows the top 15 goal scorers.")
print("2. The first player has the highest total goals.")
print("3. These players have strong goal-scoring records.")
```

# **Q no.7 : How many records are available for each season?**

```python
season_records = df["year"].value_counts()
season_records = season_records.sort_index()
print(season_records)
```

**Graph Visualization**

```python
season_records.plot(kind="line", marker="o")
plt.title("Number of Records per Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.show()
```

**Bussiness Insight**

```python
print("Business Insight:")
print("Season-wise records help compare historical data and identify data availability.")
```

**Observations**

```python
print("Observations:")
print("1. The chart shows the number of records for each season.")
print("2. Some seasons have more records than others.")
print("3. It helps identify seasons with more available data.")
```
