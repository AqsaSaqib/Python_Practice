```python
import pandas as pd
```

# **Load DataSet**

```python
df = pd.read_csv("/content/competition.csv")
df.head(10)
```

# **DataSet Information**

```python
print("Shape:\n", df.shape)
print()
print("Columns:\n", df.columns.tolist())
print()
print(df.info())
print()
print("Statistics:\n", df.describe())
```

# **Check Missing values**

```python
print(df.isnull().sum())
```

# **Handle Missing Values**

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Fare"] = df["Fare"].fillna(df["Fare"].mean())
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("After Handle missing values:\n", df.isnull().sum())
```

# **Fix any wrong datatype**

```python
num_columns = df[["Pclass", "Age", "Fare", "SibSp", "Parch", "PassengerId"]]
for i in num_columns:
    df[i] = pd.to_numeric(df[i])
print("After checking the columns:\n", df.dtypes)
```

# **Drop Duplicates**

```python
df.drop_duplicates()
print("After Drop Duplicates:\n", df.duplicated().sum())
```

# **Summary Table after groupby and aggregate**

```python
summary = (df.groupby(["Pclass", "Sex"]) .agg(P_Count=("PassengerId", "count"), Avg_Age=("Age", "mean"), Avg_Fare=("Fare", "mean"))     )
summary
```

# **New Small DataFrame**

```python
New_DataFrame= pd.DataFrame({ "Pclass": [1, 2, 3], "ClassLabel": ["First Class", "Second Class", "Third Class"]})
New_DataFrame
```

```python
final_data = summary.merge(
    New_DataFrame,
    on="Pclass",
    how="inner"
)

final_data
```
