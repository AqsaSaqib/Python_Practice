```python
# Difference between .loc[] and .iloc[]?
"""
loc[] and iloc[] can used to access the rows and columns directly
and in loc[] we use index value or label to access the data from the dataframe
but in iloc we only pass the integers value to access the data
In Slicing, loc[] the stop value is inclusive or in iloc[] stop value is exclusive
"""
```

```python
# How do you check for and count missing values in a DataFrame?
"""
we can check missing values using isnull() function and count them by using sum()
and properly we write it as isnull().sum() which gives the total missing values present
"""
```

```python
# What’s the difference between df.copy() and just assigning df2 = df?
"""
df.copy() means making the new copy of the data and any change in the copy doesn't effect the original data
df2 = df means assiging the value of df to df2 and any change in df can also cause change in df2
"""
```

```python
# How do you filter rows where two conditions are both true?
"""
we use AND or & operator to filter these rows where both conditions are true it
append the data into the filter_Arr and if the condition is false append is False.
"""
```

```python
# What does .groupby() return before you call an aggregation on it?
"""
.groupby() return a groupby data before an aggregation applies on it and when aggregation applies
it simply performs the action which is given in the function such as mean, median, sum etc
"""
```

# **Bussiness Questions**

```python
# No. of passengers in a class
df["Pclass"].value_counts().sort_index(ascending=True)
```

```python
# No. of Male and Female
df ["Sex"].value_counts()
```

```python
# average fare by each class
df.groupby("Pclass")["Fare"].mean()
```

```python
# which passengers travel with family (children, spouse, siblings, parents)
family = df[(df['SibSp'] > 0) | (df['Parch'] > 0)]
family[['Name', 'SibSp', 'Parch']]
```

```python
# passengers who pay more than avg fare
avg_fare = df["Fare"].mean()
print(avg_fare)

more_fare = df[df['Fare'] > avg_fare]
more_fare[['Name','Fare','Pclass']]
```

```python
# Passengers have missing cabins
miss_cabins = df[df["Cabin"].isnull()]
miss_cabins[["Name", "Cabin", "Pclass", "Sex"]].head(20)
```

```python
# hwhich passengers paid fare more than 150
greater_fare = df[df["Fare"]>120]
greater_fare[["Name", "Sex", "Fare", "Age"]]
```

```python
# passengers having age less than 18
age = df[df["Age"]<18]
age[["Name", "Sex", "Pclass", "Age"]]
```

```python
# how many passengers embarked from every port
df["Embarked"].value_counts()
```
